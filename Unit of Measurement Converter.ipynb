```python
convert_from = input("Enter Starting Unit of Measurement (inches, feet, yards): ")

convert_to = input("Enter Unit of Measurement to Convert to (inches, feet, yards): ")

#change data type of number of inches to interger it comes as string
#use an f string for the print output so you can combine strings and int
if convert_from.lower() in ["inches", 'inch', 'in'] :
    number_of_inches = int(input("Enter Starting Measurement in Inches: "))
    if convert_to.lower() in ["feet", 'ft', 'foot']:
        print(f" Result: {number_of_inches} inches = {round(number_of_inches / 12,2)} feet")
    elif convert_to.lower() in ["yards", 'yd', 'yard', 'yds']:
        print(f" Result: {number_of_inches} inches = {(number_of_inches / 36):.2f} yard")
    else:
        print('Your input was incorrect')
 
elif convert_from.lower() in ["feet", 'ft', 'foot']:
    number_of_feet = int(input("Enter Starting Measurement in Feet: "))
    if convert_to.lower() in ["inches", 'inch', 'in']:
        print (f" Result: {number_of_feet} feet = {number_of_feet * 12} inches")
    elif convert_to.lower() in ["yards", 'yd', 'yard', 'yds']:
        print (f" Result: {number_of_feet} feet = {(number_of_feet / 3):.2f} yards") 
    else:
        print('Your input was incorrect')

elif convert_from.lower() in ["yards", 'yd', 'yard', 'yds']:
    number_of_yards = int(input ("Enter Starting Measurement in Yards: "))
    if convert_to.lower() in ["inches", 'inch', 'in']:
        print(f" Result: {number_of_yards} yards = {number_of_yards * 36} inches")
    elif convert_to.lower() in ["feet", 'ft', 'foot']:
        print(f" Result: {number_of_yards} yards = {(number_of_yards / 3):.2f} feet")
    else:
        print('Your input was incorrect')
else:
        print('Please Enter either Inches, Feet, or Yards')


    
    
```
